```python
import pandas as pd

from conv_wm.config import get_path, load_config
from conv_wm.data.schemas import (
    EGOCOM_GROUND_TRUTH_SCHEMA,
    EGOCOM_VIDEO_INFO_SCHEMA,
)

cfg = load_config()

video_info = pd.read_csv(get_path(cfg, "egocom", "raw", "video_info"))

ground_truth = pd.read_csv(get_path(cfg, "egocom", "raw", "ground_truth"))

video_info_validated = EGOCOM_VIDEO_INFO_SCHEMA.validate(
    video_info,
    lazy=True,
)

ground_truth_validated = EGOCOM_GROUND_TRUTH_SCHEMA.validate(
    ground_truth,
    lazy=True,
)

print("video_info:", video_info_validated.shape)
print("ground_truth:", ground_truth_validated.shape)
print("Pandera EgoCom validation: OK")
```
